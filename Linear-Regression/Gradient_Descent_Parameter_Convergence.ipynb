```python
import numpy as np
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
```

```python
#Initialize parameters
m = 0
c = 0
learning_rate = 0.01
epochs = 1000
n = len(x)
# List to store MSE for each epoch
losses = []
```

```python
for i in range(epochs):
  y_pred = m * x + c
  error = y - y_pred
  # Gradients
  dm = (-2/n) * sum(x * (error))
  dc = (-2/n) * sum(error)

  # Update parameters
  m = m - learning_rate * dm
  c = c - learning_rate * dc

  # Calculate MSE and append it to losses list
  mse = np.mean(error ** 2)
  losses.append(mse)

  if i % 100 == 0:
    print(f"Epoch {i}: MSE: {mse:.4f},m: {m:.4f}, c: {c:.4f}")

```

```python
import matplotlib.pyplot as plt
```

```python
# plot the loss (MSE) vs epochs
plt.plot(range(epochs), losses, label = 'MSE loss')
plt.xlabel('Epochs')
plt.ylabel('MSE')
plt.title('Loss curve (MSE vs Epochs)')
plt.grid(True)
plt.legend()
plt.show()

#Final values
print("Final values:")
print(f"m: {m:.4f}")
print(f"c: {c:.4f}")
```
