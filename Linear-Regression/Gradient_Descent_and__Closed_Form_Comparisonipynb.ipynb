```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. Data
# -----------------------------
W = np.array([0.9, 1.8, 2.4, 3.5, 3.9, 4.4, 5.1, 5.6, 6.3])  # feature
S = np.array([1.4, 2.6, 1.0, 3.7, 5.5, 3.2, 3.0, 4.9, 6.3])  # target

# -----------------------------
# 2. Gradient Descent results (from your previous run)
# -----------------------------
# Assuming you have already normalized for gradient descent
# And final slope and intercept from gradient descent (scaled back)
M_gd =  0.727  # example, replace with your actual final M
C_gd = 0.126  # example, replace with your actual final C
S_pred_gd = M_gd * W + C_gd

# -----------------------------
# 3. Closed-form solution
# -----------------------------
M_cf = np.sum((W - W.mean()) * (S - S.mean())) / np.sum((W - W.mean())**2)
C_cf = S.mean() - M_cf * W.mean()
S_pred_cf = M_cf * W + C_cf

# -----------------------------
# 4. Plot
# -----------------------------
plt.scatter(W, S, color='blue', label='Data')
plt.plot(W, S_pred_gd, color='red', label='Gradient Descent')
plt.plot(W, S_pred_cf, color='green', linestyle='--', label='Closed-form')
plt.xlabel('Weight (W)')
plt.ylabel('Size (S)')
plt.title('Linear Regression Comparison')
plt.legend()
plt.show()
```
