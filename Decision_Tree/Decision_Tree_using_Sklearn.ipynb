```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
#Import the dataset and check it
dataset = pd.read_csv('/content/collegePlace.csv')
dataset
```

```python
dataset.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
enc = OneHotEncoder()
encoded = enc.fit_transform(dataset[['Gender']])
encoded = encoded.toarray()
encoded
```

```python
encoded = enc.fit_transform(dataset[['Stream']])
encoded = encoded.toarray()
encoded
```

```python
encoded_df = pd.DataFrame(encoded, columns=enc.get_feature_names_out(['Stream']))
```

```python
dataset_encoded = pd.concat([dataset.drop('Stream', axis = 1), encoded_df], axis=1)
dataset_encoded
```

```python
dataset_encoded = pd.concat([dataset.drop('Gender', axis = 1), encoded_df], axis=1)
dataset_encoded
```

```python
dataset_encoded = pd.get_dummies(dataset_encoded)
```

```python
X =dataset_encoded.drop('PlacedOrNot',axis=1).values
Y = dataset_encoded['PlacedOrNot'].values
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train,Y_train)
Y_pred = model.predict(X_test)
accuracy = accuracy_score(Y_test, Y_pred)
accuracy
```
