```python
import numpy as np
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Features:", iris.feature_names)
print("Classes:", iris.target_names)
```

```python
np.random.seed(42)

indices = np.random.permutation(len(X))
test_size = int(0.2 * len(X))

test_indices = indices[:test_size]
train_indices = indices[test_size:]

X_train = X[train_indices]
X_test = X[test_indices]

y_train = y[train_indices]
y_test = y[test_indices]
```

```python
X_mean = np.mean(X_train, axis=0)
X_std = np.std(X_train, axis=0)

X_std[X_std == 0] = 1.0

X_train_scaled = (X_train - X_mean) / X_std
X_test_scaled = (X_test - X_mean) / X_std
```

```python
#Calculating Eucledian Distance
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))
```

```python
def get_k_nearest_neighbors(X_train, query, k):
    distances = []

    for i in range(len(X_train)):
        distance = euclidean_distance(X_train[i], query)
        distances.append((distance, i))

    distances.sort(key=lambda item: item[0])

    nearest_indices = [
        index for distance, index in distances[:k]
    ]

    return nearest_indices
```

```python
class KNNClassifierScratch:
    def __init__(self, k=5):
        self.k = k
        self.X_train = None
        self.y_train = None

    def fit(self, X, y):
        if self.k < 1 or self.k > len(X):
            raise ValueError("k must be between 1 and the number of samples")

        self.X_train = np.asarray(X)
        self.y_train = np.asarray(y)

        return self

    def predict_one(self, query):
        indices = get_k_nearest_neighbors(
            self.X_train, query, self.k
        )

        neighbor_labels = self.y_train[indices]

        labels, counts = np.unique(
            neighbor_labels, return_counts=True
        )

        return labels[np.argmax(counts)]

    def predict(self, X):
        return np.array([
            self.predict_one(query) for query in X
        ])
```

```python
def accuracy_score_scratch(y_true, y_pred):
    return np.mean(y_true == y_pred)


knn = KNNClassifierScratch(k=5)

knn.fit(X_train_scaled, y_train)

y_pred = knn.predict(X_test_scaled)

print("K:", knn.k)
print("Test accuracy:", accuracy_score_scratch(y_test, y_pred))
print("Correct predictions:", np.sum(y_test == y_pred))
print("Total test samples:", len(y_test))
```

```python
for i in range(len(X_test)):
    actual = iris.target_names[y_test[i]]
    predicted = iris.target_names[y_pred[i]]

    print(
        f"Actual: {actual:12s} | "
        f"Predicted: {predicted}"
    )
```
